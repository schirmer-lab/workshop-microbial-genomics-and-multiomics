### 3. Data types and data structures ###

Variables are identifiers with names that you can use to store values etc. These can then be called upon later, to examine or use their stored values. The assignment can be done using **<-** or **=**

```r
x <- 2 # x = 2 also works
y <- 3

# x + y

#The result can be stored in another variable

z <- x + y
# z
```

These variables are of type integers. Some example of data types in R are:

* numeric:
    * integer: 5L (L tells R to store this as an integer)
    * double: 33.6
* character(s): "g", "R", "Fun", "games"
* logical: TRUE, FALSE


To check the datatype of a variable:

```r
x <- 3L  #L tells R to store this as an integer. The value stored in x is 3
typeof(x)

y <- 9.4
typeof(y)

my_name <- "Mia"     #Remember the quotes for characters
typeof(my_name)

my_logical_var <- TRUE
typeof(my_logical_var)
```


These examples above show variables holding only one value. There are basic data structures in R that can hold multiple values.

* Vectors
* Matrices
* Data frames
* Factors
* Lists


```r
my_vector1 <- c(11:16)          #c() is an internal R function to combine values;
my_vector1

my_vector2 <- seq(51, 56)       #seq() is an internal R function to generate a sequence of values
my_vector2

my_vector3 <- c("a", "b", "c", "d", "e", "f")
my_vector3

#Accessing values of a vector using indices

my_vector1[1]
my_vector1[c(1,2,5)]

#To check if a value is present in the vector
"d" %in% my_vector3
12 %in% my_vector1

#To get also the position of the matching value(s)
which(my_vector3 %in% "d")
```

**Matrices in R**

They can only store one data type, so all values in the matrix are either integers, characters, or something else.


```r
my_matrix1 <- matrix(3, nrow = 2, ncol = 2)
my_matrix1

my_matrix2 <- matrix(my_vector1, nrow=2, ncol=3, byrow=TRUE)
my_matrix2

my_matrix3 <- cbind(my_vector2, my_vector3)
my_matrix3

#Accessing a row/column/individual value
my_matrix3[1,]      #1st row
my_matrix3[,1]      #1st column
my_matrix3[1,1]     #1st row and 1st column

```

In the case of my_matrix3, all data is converted into character. What if we want to combine different datatype into one object?




**Dataframes in R**

When we want to store more than one datatype in the same object we can use dataframes.


```r
my_df1 <- data.frame(my_vector2, my_vector3)
my_df1

#compare with my_matrix3
my_matrix3

#Making another dataframe with several different datatypes
v1 <- c(1:10)                                         #Integer
v2 <- rnorm(1:10)                                     #Double
v3 <- letters[1:10]                                   #Character
v4 <- rep(c(T,F),5)                                   #Logical
v5 <- c(rep("Healthy", 5), rep("Diseased", 5))        #Character

my_df2 <- data.frame(v1, v2, v3, v4, v5)
my_df2
```

While different datatypes are presevered in dataframes, all entries need to be of the same length.


**Factors in R**  

Factors can store categorical data.

```r
#lets convert column 5 in my_df2 to factors
my_df2[,5] <- as.factor(my_df2[,5])
my_df2[,5]
```

**Lists in R**  

Lists can also hold various data structures and different datatypes. Let's make a list with a vector, a matrix, a dataframe, and another list.


```r
my_list1 <- list("R-Workshop", "Awesome day", c(4,58), 5.3, 1001.3, TRUE)
my_list1
my_list2 <- list(my_vector1, my_matrix1, my_df1)
my_list2

#Accessing elements in the list through indices
my_list2[[1]]
my_list2[[1]][1]
```

Defining rownames and columnnames for matrices and dataframes

```r
rownames(my_matrix1)
colnames(my_matrix1)

rownames(my_df1)
colnames(my_df1)

names(my_list2)
```

We can add/change rownames, columnnames and names of your data structure


```r
#Adding rownames and colnames to the matrix
rownames(my_matrix1) <- c(paste("r", c(1:2), sep=""))
rownames(my_matrix1)
my_matrix1

#Adding rownames and colnames to the dataframe
colnames(my_df1) <- c(paste("vec", c(1:2), sep=""))
colnames(my_df1)
my_df1
```


### 4. Functions ###

A function is a set of statements organised together to perform specific tasks. R has a large number of built-in functions (examples: c(), print(), paste(), seq(), sum(), min(), max(), ...) and users can create their own functions. Let us explore a few built-in functions.


```r
sum(1:10)
min(3,4,7,2,9,11,56,10)
mean(33,51,16,5,4,98, 10, 92)

```

To get help for any function type the function name preceded by a '?'

```r
?sum()
```

The syntax to create our own function:

```r
#Not to run
function_name <- function(arg_1, arg_2, ...) {
   command1
}
```

Let's write our first R function that executes a simple arithmetic operation and then returns the resulting value.


```r

eval_function <- function(num1, num2, num3, num4)
{
    result <- (num1 + num2)*(num3 - num4)
    return(result)
}
```

Now that the function is created, let's call the function with varying arguments.

```r
eval_function(num1 = 11, num2 = 53, num3 = 72, num4 = 37)

#changing the order of the arguments does not change the output if they are passed by name
eval_function(num3 = 72, num1 = 11, num4 = 37, num2 = 53)

#Arguments are passed by position
eval_function(11, 53, 72, 37)

#Saving the returned value to a variable
res <- eval_function(11, 53, 72, 37)
print(res)
```

An important class of functions is the **apply()** family, such as apply(), lapply(), sapply() and lapply(). To get help on a function try *?apply()*, *?sapply()*, ...

Some examples:


```r
set.seed(100) #To make the answer reproducible, we set the seed for the random numbers generation.

#generate a 5x6 matrix of random numbers between 1-100
my_mat1 <- matrix(sample(1:100, 30, replace=T), nrow=5, ncol=6)
my_mat1

#apply()
my_sum1 <- apply(my_mat1, 1, function(x) sum(x))      #1 implies row-wise operation
my_sum1

my_sum2 <- apply(my_mat1, 2, function(x) sum(x))      #2 implies column-wise operation
my_sum2

#lapply()
lapply(my_mat1[1,], function(x) sum(x))
lapply(my_mat1[,1], function(x) sum(x))

#sapply()
sapply(my_mat1[,1], function(x) sum(x))
sapply(my_mat1[1,], function(x) sum(x))
```

Did you check how the output data's structure differs for all the functions?

A popular package included in tidyverse is **ggplot2**. It is a data visualization package.


```r
# install.packages("reshape2")
# install.packages("ggplot2")

library("reshape2") # melt function
library("ggplot2")

data <- data.frame(
  name=c("A","B","C","D","E") ,
  value=c(3,12,5,18,45)
  )
data

# Barplot
ggplot(data, aes(x=name, y=value)) +
  geom_bar(stat="identity")
```

```r
#Point plots
dataframe2 <- data.frame(
  name=c("A","B","C","D","E") ,
  value1=c(3,12,5,18,45),
  value2=c(10,6,11,9,19)
  )
print(dataframe2)

ggplot(dataframe2, aes(x=value1, y=value2)) +
  geom_point(color="blue", shape=4, size=6)
```

```r
set.seed(100)       #For reproducibility of random number generation

dataframe3 <- data.frame(
  value1=sample(0:250, 10, replace = FALSE),
  value2=sample(0:150, 10, replace = TRUE)
)
dataframe3

#melt is used to reshape and elongate the data frames
dataframe_reshaped <- melt(dataframe3)
dataframe_reshaped

# Creating a box plot
ggplot(dataframe_reshaped, aes(x=variable, y=value)) +
   geom_boxplot(color="black", fill=c("red","blue"), alpha=0.9) +
   theme_bw()+
   theme(axis.text=element_text(size=20),axis.title=element_text(size=18,face="bold"))
```

Check what different functions are available in the R-library/package

```r
# Check what different functions are available in the R-library/package
# Note: If you get a "lazy-load database is corrupt" error, the package needs to be reinstalled

tryCatch({
  lsf.str("package:ggplot2")
}, error = function(e) {
  cat("Error accessing ggplot2 functions:\n")
  cat(e$message, "\n\n")
  cat("This indicates a corrupted package installation.\n")
  cat("To fix this, reinstall ggplot2:\n")
  cat("remove.packages('ggplot2')\n")
  cat("install.packages('ggplot2')\n\n")

  # Try alternative approach to list functions
  cat("Alternative: Using ls() to show available objects in ggplot2:\n")
  if ("ggplot2" %in% loadedNamespaces()) {
    cat("ggplot2 namespace is loaded. Some key functions include:\n")
    cat("ggplot, aes, geom_point, geom_bar, geom_line, theme, etc.\n")
  } else {
    cat("ggplot2 namespace not loaded. Try: library(ggplot2)\n")
  }
})
```

```r
# If you encountered the "lazy-load database is corrupt" error above,
# run this cell to fix the ggplot2 installation:

fix_ggplot2 <- function() {
  cat("Fixing corrupted ggplot2 installation...\n")

  # Remove the corrupted package
  try(remove.packages("ggplot2"), silent = TRUE)

  # Clean up any remaining files
  lib_path <- .libPaths()[1]
  ggplot2_path <- file.path(lib_path, "ggplot2")
  if (dir.exists(ggplot2_path)) {
    cat("Removing corrupted ggplot2 directory...\n")
    unlink(ggplot2_path, recursive = TRUE)
  }

  # Reinstall ggplot2
  cat("Reinstalling ggplot2...\n")
  install.packages("ggplot2", dependencies = TRUE)

  # Test the installation
  library(ggplot2)
  cat("✅ ggplot2 reinstalled successfully!\n")

  # Now try to list functions again
  cat("\nTesting function listing:\n")
  lsf.str("package:ggplot2")
}

# Uncomment the line below to run the fix:
# fix_ggplot2()
```
