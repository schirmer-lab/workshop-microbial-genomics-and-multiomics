- **Experimental**: RCSB PDB (https://www.rcsb.org)

## What can Foldseek do?

Foldseek allows fast and accurate comparison of protein structures using the so-called 3Di sequence as well as the amino acid sequence.

### 🛠️ Steps:
1. **Create a Foldseek database** from your set of PDB files.
2. **Run pairwise comparisons** between structures.
3. **Interpret the results** to identify structural similarities.

### 🧪 Special Task:
You can also use **ProtT5** to generate embeddings from amino acid sequences and compare them if you have no structure file available or work with large datasets.


## 3. How to Analyze Individual Protein Structures (ChimeraX)

ChimeraX is a powerful tool for visualizing and analyzing protein structures.

### 🔍 Basic Usage:
- **Fetch structures** by PDB ID
- **Manipulate view**: rotate, zoom, select atoms/residues

### 🔗 Interaction Analysis:
- Hydrogen bonds
- Van der Waals contacts

### 🔄 Structure Comparison:
- RMSD calculation
- MatchMaker tool for alignment

#### 📝 Task:
- Open ChimeraX and fetch structure `5TOA`
- Identify the ligand bound to the receptor
- Interaction & binding analysis:
    - Hide chain B including the estradiol ligand
    - Use Tools > Structure analysis > H bonds and Contacts to visualize interactions between the receptor and estradiol
    - Identify hydrogen bonds and Van der Waals contacts: color all involved amino acids magenta, display side chains as sticks and color by heteroatom
- Structure comparison:
    - Load a second structure (e.g. mouse estrogen receptor or human estrogen receptor alpha)
    - Use Tools > Structure Analysis > Matchmaker to align the second structure to the first
    - Align the two structures using the command line: match #2 to #1
    - Check the RMSD value in the log panel
    - Calculate RMSD for the binding region only by modifying the following expression "match #1/A:10-50 to #2/A:10-50"
