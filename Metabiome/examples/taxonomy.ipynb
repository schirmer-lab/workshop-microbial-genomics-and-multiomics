### Build NCBI taxonomy database (assumed you have already downloaded the NCBI taxonomy database)

```python
# You'll have to  pre-download NCBI taxonomy database in .tar.gz format, you can specify the output file name,
# but you'll have to only use this name for mapping NCBI to GTDB as well as MetaPhlAn
build_lineage(taxa=parse_ncbi_taxonomy("../tmp/taxdump.tar.gz"), out="../tmp/taxa.csv")
```

### Convert GTDB-Tk and MetaPhlAn taxonomy database to NCBI taxonomy database

```python
process_gtdb_annotations(
    gtdb="../data/GTDB",
    ncbi="../tmp/taxa.csv",
    out="../output",
)  # Note, there are multiple tsv files from GTDB, parse a directory without "/" suffix to process all tsv files for a complete database
```

```python
process_metaphlan_annotations(
    metaphlan="../data/metaphlan/mpa_v30_CHOCOPhlAn_201901_marker_info.txt",
    ncbi="../tmp/taxa.csv",
    out="../output",
)  # You can specify the output name of mapped MetaPhlAn or GTDB taxonomy database in src.annnotations.taxonomy
```
